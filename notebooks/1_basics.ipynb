```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)
```

```python
model.invoke("what is the temp in la")
```

## **first graph**

### **step 1 define the schmea**

```python
from typing import TypedDict , List

class graph_Schema(TypedDict):
    name: str
    age: int
    message: str
```

### create the node func

```python
def welcome(state:graph_Schema)->graph_Schema:

    curr_name = state['name']
    curr_age = state['age']
    curr_message = state['message']

    response = model.invoke(f"hello {curr_name} your age is {curr_age} and your message is {curr_message}")
    response = response.content[0]['text']
    state['message']= f"your message was {curr_message}. here is your response {response}"

    return state
```

```python
from langgraph.graph import StateGraph , START,END

graph = StateGraph(graph_Schema)#graph bobject
#adding nodes
graph.add_node("welcome",welcome)
#Adding edges
#these are default nodes start and end
graph.add_edge(START,"welcome")
graph.add_edge("welcome",END)
```

### step 4 compile the graph

```python
from IPython.display import display,Image





first_graph = graph.compile()#graph is the obj we created earlier


#you could see the error with the below command
Image(first_graph.get_graph().draw_mermaid_png())


#best command
# print(first_graph.get_graph().draw_mermaid())
```

### step 5 run the graph

```python
first_graph.invoke({"name":"tushar","age":22,"message":"hello"})
```
